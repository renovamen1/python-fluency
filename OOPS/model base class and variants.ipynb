## Model Base Class and Variants

Inheritance and polymorphism — a family of "models" that all speak the same interface.

Model (base class):

<li>name</li>
<li>nameis_trained</li>
<li>train(data)</li>
<li>redict(data)</li>

LinearModel(Model):

<li>weights</li>
<li>override train() and predict()</li>

ThresholdClassifier(Model):

<li>threshold</li>
<li>override train() and predict()</li>

Requirements:

<li>predict() must refuse to run (with a clear message, not a crash) if train() hasn't been called yet.</li>
<li>Put several different model instances in one list and call .predict() on each in a loop, using the same line of code for every one of them, regardless of which subclass it actually is.</li>
<li>Each subclass's train() should set is_trained = True only after doing something with the data (doesn't need to be real math yet — just needs to behave like a distinct step happened).</li>
