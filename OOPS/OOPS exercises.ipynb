### Exercise 1: Student Class
Create a Student class.<br>
Requirements:
<li> Attributes: name, age, course</li>
<li>Create a method introduce() that prints the student's details.</li>
<li>Create two different Student objects.</li>
<li> Call introduce() for both.</li>

```python
class Student:
    def __init__(self,name,age,course):
        self.name = name
        self.age = age
        self.course = course
    def introduce(self):
        print("Student Details:")
        print(f"Name:{ self.name}\nAge: {self.age}\nCourse: {self.course}\n")
        
```

### Exercise 2: Bank Account
Create a BankAccount class.<br>
Requirements:<br>
<li> Attributes: account_holder, balance</li>
<li> Method deposit(amount)</li>
<li> Method withdraw(amount)</li>
<li> Method show_balance()</li>
<li>Do not allow withdrawal if the balance is insufficient.</li>

```python
class BankAccount:
    def __init__(self,acc_holder,bal):
        self.accout_holder = acc_holder
        self.balance = bal
    def deposit(self,amount):
        self.balance = self.balance + amount
        print(f"Rs.{amount} has been deposited to your account. your current balance is Rs.{self.balance}")
    
    def withdraw(self,amount):
        if (self.balance > amount):
            self.balance = self.balance - amount
            print(f"Rs.{amount} has been withdrawn from your account. your current balance is Rs.{self.balance}")
        else:
            print("Insufficient Balance.")

```

```python
# oops., this was lot of work to get this right. took me whole damn 3 days to get this right.
# and i am just super proud to get this done and working on my own, good work kid!

# this exercise was suuper duper helpful to feel the need of 'break' and 'continue' statments. this is a jackpot for going through those.
# must visit it later to polish the concepts. 
class Book:
    def __init__(self,title,author,is_available):
        self.title = title
        self.author = author
        self.is_available = is_available

class Library(Book):
    def __init__(self):
        self.books = [{'title':'shoedog','author':'philip knight','available':True}]
    def add_book(self,title,author,is_available = True):
        super().__init__(title,author,is_available)
        if (len(self.books) == 0):# simply append /add the book.
            self.books.append({'title':title.lower(),'author':author.lower(),'available':is_available})
        else:
            for itr in range(len(self.books)):
                if (title.lower() == self.books[itr]['title']): # if the desired book already exist, we simply break out of the loop
                    print("This Book already exists.")
                    break
                else: 
                    if(itr < (len(self.books)-1)):
                        continue  # it skips the remaining portion of the current iteration in a loop and takes back to the condiion statement. CONTINUE cannot increment the iterator, it just takesit back to the condition.
                    else: 
                        self.books.append({'title':title.lower(),'author':author.lower(),'available':is_available})
    def show_books(self):
        if (len(self.books) == 0):
            print("the library is empty")
        else:
            for itr in self.books:
                print(itr)
    def borrow_book(self,title):
        if (len(self.books) == 0):
            print("ther are no book in the library to borrow.\n")
        else:
            for book in range(len(self.books)):
                if ( (title.lower() == self.books[book]['title']) and (self.books[book]['available'] == True) ):
                    self.books[book]['available'] = False
                    print(f"This Book '{title}' has been borrowed successfully.")
                    break # it just exists the execution out of the loop
                elif ( (title.lower() == self.books[book]['title']) and (self.books[book]['available'] == False) ):
                    print(f"the book {title} is currently unavailable")
                    break
                elif ( (title.lower() != self.books[book]['title']) and (book != (len(self.books)-1))):
                    continue
                else:
                    print(f"there is no book in the library with the name '{title}'.")
    def return_book(self,title):
        for book in range(len(self.books)):
            if (title.lower() == self.books[book]['title']):
                if (self.books[book]['available'] == False):
                    self.books[book]['available'] = True
                    print(f"The book '{title}' has been returned successfully")
                    break
                else:
                    print(f"this book{title} is already returned and is available to borrow.")
                    break
            elif (book == (len(self.books)-1)):
                print(f" The book you are trying to return '{title}' doesn't exist.")
            else:
                continue
```
