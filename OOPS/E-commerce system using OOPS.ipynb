```python
class Product:
    def __init__(self,name,price,quantity):
        self.name = name.lower()
        self.price = price
        self.quantity = quantity

class collections:
    def __init__(self):
        self.items = []

class Store: # this is like a store's collection of products they sell.
    def __init__(self):
        self.products = collections()   #  class attribute errors when multiple instances of 'Store' end up sharing same 'products' class attribute. it is available across all instances of class but different stores share different products and collection, so changing it to instance attribute 
    def add_product(self,name,price,stock):
        new_product = Product(name,price,stock)
        self.products.append(new_product)
    def remove_product(self,name):
        if (len(self.products) == 0):
            print("The store has no product available.")
        else:
            for item in self.products:
                if (item.name == name.lower()):
                    self.products.remove(item)
                    return
                else:
                    continue
            print(f"The product '{name}' you are trying to remove doesn't exist in the inventory.")
    def show_products(self):
        if (len(self.products)== 0):
            print("The store is empty. Please add products in the inventory.")
        else:
            print("Products".ljust(10),"Price".ljust(5),"Stock".rjust(4))
            for item in self.products:
                print(item.name.ljust(10),item.price,"\t",item.quantity)

# cart is not a product, cart contains product.
class Cart: # so,we are not inhereting 'Product' class here but using composition
    def __init__(self):
    # the items in the list should be real instances of products
        self.cart_items = [] # here i have initialized an empty list
        # it would contain instances of class Product or simply products added to the cart with thier details

    def add_product(self,name,purchase_quantity):
        for item in self.cart_items:
            if (name.lower() == item.name):
                print("the product is already added to the cart.\n please remove the exising product and add product again to increase the purchase quantity. ")
                return
        for store_item in Store.products:
            if ( name.lower() == store_item.name):
                if  (store_item.quantity == 0):
                    print(f"The product '{store_item.name}' is Out of stock!")
                    return
                elif ( purchase_quantity > store_item.quantity):
                    print(f"the quantity of '{name}' exceeds the available stock. ")
                    print(f" As per current stock, you can purchase maximum {(store_item.quantity)} '{store_item.name}' products.")
                    return
                else:
                    price = store_item.price # by doing so, we dont require to pass price when adding a product to a cart and avoids complications regarding product prices.
                    picked_product = Product(name,price,purchase_quantity) # creating new instances of product and appending them to the list
                    self.cart_items.append(picked_product)
                    break
        else: # we can use a else block to a for loop, this else branch of for loop is only executed if the loop runs smoothly with out encountering a break
            print(f"the product '{name} you are trying to add to the cart doesn't exist in store products collection.") # if a break is encountered in the body of for loop, the else part of the for loop is automatiacally skipped.

    def remove_product(self,product_name):
        for  item in self.cart_items:
            if (item.name == product_name.lower()): # whatever the products added to store are already in lower(). 
                self.cart_items.remove(item)
                print(f"the '{product_name}' is removed from the cart" )
                return
            else:
                continue
        print(f"The product '{product_name}' you are trying to remove cannot be found.")

    def calculate_total(self):
        sum = 0
        for item in self.cart_items:
            sum = sum + (item.price * item.quantity)
        return sum
    def show_cart(self):
        if (len(self.cart_items)== 0):
            print("The cart is empty. Please add products to your cart.")
        else:
            print("Products".ljust(10),"Price".ljust(5),"Quantity".rjust(4))
            for item in self.cart_items:
                print(item.name.ljust(10),item.price,"\t",item.quantity) 
                '''
                the problem below: has been fixed. i simply renamed the 'stock' instance attribute of 'Product' class as 'quantity' to avoid conflict and reflect the changes in other areas.thus  issues of using name 'stock' to represent even the 'purchase_quantity' is resolved. 
                # since 'Product' object has no attribute 'purchase_quantity'
                # it is storing 'purchase_quantity' and gives the same result we want but it is just using name as 'stock' only even tho we intend or mean to use the purchase_quantity of the product 
                # we are doing it because even tho we add a product in cart using  'purchase_quantity', the add_product() method is ultimately making instance of the class 'Product'. thus product only recognizes stock
                '''
# customer is not cart(so,we wont inherit.)
# customer contains cart. so we create instance of cart inside it(compositon!).
class Customer():
    def __init__(self,name):
        self.name = name
        self.cart = Cart()

    def checkout(self):
        for cart_item in self.cart.cart_items:
            for store_item in Store.products:
                if (cart_item.name == store_item.name):
                    if (store_item.quantity >= cart_item.quantity):
                        store_item.quantity = store_item.quantity - cart_item.quantity
                    # i have already added this checks for orders with stock limits while adding producuts in the cart,so every product added to cart now stays wiht in the stock limit.
                    #     print(f"the ordered quantity of product '{cart_item.name}' exeeds the available stock.")
                    #     print(f" As per current stock, you can purchase maximum {(store_item.quantity)}'{cart_item.name} ")
            


        print("Customer Details \n \n")
        print("Name: ",self.name)
        self.cart.show_cart()
        print("\n \n")
        
        if (self.cart.calculate_total() > 5000):
            print(f"Total Price Of Products in the Cart is Rs.{self.cart.calculate_total()}")
            print("Since your purchase exceed Rs.5000. you recieve 10% discount.")
            print(f"Total Price Of Products in the Cart after discount is Rs.{(self.cart.calculate_total() - (0.10 * self.cart.calculate_total()))}")
        else:
            print(f"Total Price Of Products in the Cart: is Rs.{self.cart.calculate_total()}")

        
```
