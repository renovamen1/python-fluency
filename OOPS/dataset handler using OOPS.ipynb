## Dataset Handler

Encapsulation basics — the OOP equivalent of a tiny DataFrame you build and control yourself.

Dataset:

<li>records (a list of dicts, each dict = one row)
<li>feature_names<l/i>
<li>add_record()</li>
<li>remove_record(index)</li>
<li>get_column(feature_name)</li>
<li>summary()</li>

Requirements:

<li>Every record added must have exactly the same keys as feature_names — reject and explain if not.</li>
<li>get_column() returns all values for one feature across every record.</li>
<li>summary() prints record count, feature names, and the mean of any numeric feature (skip non-numeric ones cleanly).</li>
<li>Some values might be missing (None) — decide how the mean calculation should handle that, and make sure it doesn't crash.</li>
